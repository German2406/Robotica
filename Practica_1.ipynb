**1. Definición de URDF**

*Unified Robot Description Format*, por sus siglas en inglés es un formato de archivo basado en XML utilizado en ROS (Robot Operating System) para describir todos los elementos y propiedades físicas de un robot. Define la estructura geométrica y cinemática del robot mediante la interconexión de eslabones rígidos  y articulaciones. Asimismo, especifica propiedades físicas fundamentales como masas, matrices de inercia, centros de masa, límites de movimiento y archivos de malla visuales para la representación gráfica en 3D.

**2. Tipos de juntas definidas dentro del formato URDF**

El estándar URDF soporta varios tipos de articulaciones para modelar el movimiento relativo entre los eslabones:

- **Revolute** (revolución)**:** Articulación de rotación con límites estrictos definidos por un rango angular mínimo y máximo.
- **Continuous** (continua)**:** Articulación de rotación que puede girar indefinidamente sin límites angulares fijos (común en ruedas).
- **Prismatic** (prismatica)**:** Articulación de traslación (lineal), donde el eslabón se desliza a lo largo de un eje con límites de desplazamiento.
- **Fixed** (fija)**:** Unión rígida que no permite ningún grado de libertad ni movimiento relativo entre los eslabones conectados.
- **Floating** (flotante)**:** Articulación que permite libertad total de movimiento en el espacio tridimensional (6 grados de libertad: 3 de translación y 3 de rotación).
- Planar (Planar)**:** Articulación que permite movimiento en un plano bidimensional de 3 grados de libertad (dos traslaciones y una rotación).

**3. Descripción de los bloques que integran la simulación de un robot en Simulink y Simscape**

Cuando se importa un archivo URDF a un entorno de Simulink mediante Simscape Multibody, el modelo resultante se compone de los siguientes bloques principales:

- **Bloque World** (Mechanism Configuration)**:** Es el entorno de referencia inercial absoluto que fija el sistema coordenado global del mundo y define parámetros físicos globales como la gravedad y las unidades de medida.
- **Bloques de Cuerpos (**base\_link **,** link\_1 **,** link\_2 **,** etc.**):** Representan los eslabones rígidos del robot. Contienen la información de masa, inercia y la geometría visual/de colisión importada del URDF.
- **Bloques de Unión:** Modelan las articulaciones físicas que conectan un cuerpo con otro. Permiten configurar el tipo de movimiento, restricciones, límites mecánicos, así como habilitar puertos de actuación externa y de sensado.
- **Bloques de Actuación y Control** (Entradas)**:** Permiten inyectar señales externas (como posiciones articulares calculadas por cinemática inversa o trayectorias polinómicas) hacia los puertos de movimiento (Motion) de las uniones.
- **Bloques de Sensores y Recolección de Datos:** Extraen variables físicas en tiempo real desde los puertos de las uniones—tales como la posición angular real (q), velocidad (w) y par/esfuerzo mecánico (t)—para su procesamiento o registro en el espacio de trabajo de MATLAB.

1. Se generarará la simulación del robot en SimScape a partir del archivo URDF.

5. Se importará la información de la solución de la cinemática inversa del robot del espacio de trabajo en Simulink a la simulación en SimSCape.

## Conclusiones

En conclusión, la implementación de la cinemática inversa en el modelo SCARA dentro de Simulink y Simscape permitió validar el comportamiento del robot a partir de objetivos espaciales y trayectorias articulares suaves a lo largo de los $10\;\textrm{s}$, comprobando que las señales de posición y velocidad coinciden con el caso directo cuando se conservan las mismas condiciones, mientras que el par y la potencia conservan la misma tendencia general pero muestran picos más altos y espigas asociadas al paso de la solución cinemática, el muestreo de señales y la dinámica propia del modelo.  La cinemática directa e inversa determinan, respectivamente, la posición/orientación del efector final a partir de las variables articulares y las coordenadas articulares requeridas para alcanzar un objetivo espacial; sin embargo, los esfuerzos mecánicos ( $\tau$ ) y la potencia ( $P=\tau \cdot \omega$ ) dependen de la dinámica del sistema, masas, inercias, aceleraciones y configuración física en Simscape, por lo que trayectorias de posición y velocidad equivalentes pueden acompañarse de picos distintos de par y potencia.
